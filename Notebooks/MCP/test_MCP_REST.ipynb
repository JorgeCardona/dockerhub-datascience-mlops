```bash
import nest_asyncio
import asyncio
from mcp.client.sse import sse_client
from mcp.client.session import ClientSession

# Habilita el event loop anidado en Jupyter
nest_asyncio.apply()

async def run_mcp_client():
    url = "https://calculadora-mcp.onrender.com/mcp"

    async with sse_client(url) as (read_stream, write_stream):
        async with ClientSession(read_stream, write_stream) as session:
            await session.initialize()

            # 1. Listar herramientas
            tools_response = await session.list_tools()
            print("Herramientas disponibles:")
            for tool in tools_response.tools:
                print(f" - {tool.name}: {tool.description}")

            print("\n----------------------------\n")

            # 2. Probar llamada a la herramienta 'add'
            result = await session.call_tool(
                name="add",
                arguments={
                    "operation": {
                        "a": 5.0,
                        "b": 10.0
                    }
                }
            )

            print("Resultado de la operación 'add':")
            print(result.content)

# En JupyterLab ejecutas la función asíncrona directamente con await:
await run_mcp_client()
```
